```python
import numpy as np
import pandas as pd
import joblib
```

```python
model = joblib.load('../models/linear_regression_model.pkl')
scaler = joblib.load('../models/feature_scaler.pkl')
encoder = joblib.load('../models/categorical_encoder.pkl')
```

```python
raw_user_input = {
    'Present_Price': 10.0,
    'Kms_Driven': 45000,
    'Owner': 0,
    'Car_Age': 6,
    'Fuel_Type': 'Diesel',
    'Seller_Type': 'Individual',
    'Transmission': 'Manual'
}
```

```python
input_df = pd.DataFrame([raw_user_input])
```

```python
input_df['log_present_price'] = np.log1p(input_df['Present_Price'])
```

```python
cat_cols = ["Fuel_Type", "Seller_Type", "Transmission"]

encoded_cat_array = encoder.transform(input_df[cat_cols])
encoded_cat_cols = encoder.get_feature_names_out(cat_cols)
df_encoded_cat = pd.DataFrame(encoded_cat_array, columns = encoded_cat_cols, index=input_df.index)
```

```python
scale_cols = ["Kms_Driven", "Car_Age", "log_present_price"]

input_df_scaled = input_df.copy()
input_df_scaled[scale_cols] = scaler.transform(input_df[scale_cols])
```

```python
X_inference = pd.concat([input_df_scaled, df_encoded_cat], axis=1)
X_inference.drop(columns = cat_cols + ['Present_Price'], inplace=True)
features_list = ['Kms_Driven',
 'Owner',
 'log_present_price',
 'Car_Age',
 'Fuel_Type_Diesel',
 'Fuel_Type_Petrol',
 'Seller_Type_Individual',
 'Transmission_Manual']
X_inference = X_inference[features_list]
```

```python
log_prediction = model.predict(X_inference)[0]
print(log_prediction)
```

```python
final_price_lakhs = np.expm1(log_prediction)
print(f"Final Predicted Resale Value: {final_price_lakhs}")
```
